```python
import pandas as pd
import numpy as np

# ============================================================
# STEP 1: Load Dataset
# ============================================================

file_path = "Cleaned_Dataset.xlsx"

df = pd.read_excel(file_path)

print("Dataset Loaded Successfully")
print("Shape:", df.shape)


# ============================================================
# STEP 2: View First 5 Rows
# ============================================================

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# STEP 3: View Last 5 Rows
# ============================================================

print("\nLast 5 Rows:")
print(df.tail())


# ============================================================
# STEP 4: Dataset Information
# ============================================================

print("\nDataset Information:")
print(df.info())


# ============================================================
# STEP 5: Check Missing Values
# ============================================================

missing_values = df.isnull().sum()

print("\nMissing Values:")
print(missing_values)


# ============================================================
# STEP 6: Check Duplicate Rows
# ============================================================

duplicate_rows = df.duplicated().sum()

print("\nDuplicate Rows:", duplicate_rows)


# ============================================================
# STEP 7: Basic Statistical Summary
# ============================================================

statistical_summary = df.describe()

print("\nStatistical Summary:")
print(statistical_summary)


# ============================================================
# STEP 8: Total Jobs Posted
# ============================================================

total_jobs = df["job_id"].nunique()

print("\nTotal Jobs Posted:", total_jobs)


# ============================================================
# STEP 9: Total Companies
# ============================================================

total_companies = df["company_name"].nunique()

print("\nTotal Companies:", total_companies)


# ============================================================
# STEP 10: Job Status Analysis
# ============================================================

job_status_analysis = (
    df["job_status"]
    .value_counts()
    .reset_index()
)

job_status_analysis.columns = [
    "job_status",
    "job_count"
]

print("\nJob Status Analysis:")
print(job_status_analysis)


# ============================================================
# STEP 11: Jobs by Location
# ============================================================

location_analysis = (
    df.groupby("location")
    .agg(
        job_count=("job_id", "nunique"),
        average_salary_lpa=("salary_lpa", "mean")
    )
    .reset_index()
    .sort_values("job_count", ascending=False)
)

location_analysis["average_salary_lpa"] = (
    location_analysis["average_salary_lpa"].round(2)
)

print("\nJobs by Location:")
print(location_analysis)


# ============================================================
# STEP 12: Jobs by Company
# ============================================================

company_analysis = (
    df.groupby("company_name")
    .agg(
        job_count=("job_id", "nunique"),
        average_salary_lpa=("salary_lpa", "mean")
    )
    .reset_index()
    .sort_values("job_count", ascending=False)
)

company_analysis["average_salary_lpa"] = (
    company_analysis["average_salary_lpa"].round(2)
)

print("\nJobs by Company:")
print(company_analysis)


# ============================================================
# STEP 13: Jobs by Job Title
# ============================================================

job_title_analysis = (
    df.groupby("job_title")
    .agg(
        job_count=("job_id", "nunique"),
        average_salary_lpa=("salary_lpa", "mean"),
        average_experience=("experience_required", "mean")
    )
    .reset_index()
    .sort_values("job_count", ascending=False)
)

job_title_analysis["average_salary_lpa"] = (
    job_title_analysis["average_salary_lpa"].round(2)
)

job_title_analysis["average_experience"] = (
    job_title_analysis["average_experience"].round(2)
)

print("\nJobs by Job Title:")
print(job_title_analysis)


# ============================================================
# STEP 14: Monthly Job Posting Analysis
# ============================================================

df["month"] = df["jobs_posted_date"].dt.month
df["month_name"] = df["jobs_posted_date"].dt.strftime("%B")

monthly_analysis = (
    df.groupby(["month", "month_name"])
    .agg(
        job_count=("job_id", "nunique")
    )
    .reset_index()
    .sort_values("month")
)

print("\nMonthly Job Posting Analysis:")
print(monthly_analysis)


# ============================================================
# STEP 15: Event Status Analysis
# ============================================================

event_status_analysis = (
    df["event_status"]
    .value_counts()
    .reset_index()
)

event_status_analysis.columns = [
    "event_status",
    "event_count"
]

print("\nEvent Status Analysis:")
print(event_status_analysis)


# ============================================================
# STEP 16: Event Success Rate
# ============================================================

total_events = len(df)

successful_events = (
    df["event_status"]
    .eq("success")
    .sum()
)

event_success_rate = (
    successful_events / total_events * 100
)

event_success_summary = pd.DataFrame({
    "metric": [
        "Total Events",
        "Successful Events",
        "Event Success Rate"
    ],
    "value": [
        total_events,
        successful_events,
        round(event_success_rate, 2)
    ]
})

print("\nEvent Success Summary:")
print(event_success_summary)


# ============================================================
# STEP 17: Jobs by Source System
# ============================================================

source_analysis = (
    df["source_system"]
    .value_counts()
    .reset_index()
)

source_analysis.columns = [
    "source_system",
    "job_count"
]

print("\nJobs by Source System:")
print(source_analysis)


# ============================================================
# STEP 18: Skill Threshold Analysis
# ============================================================

skill_threshold_analysis = (
    df.groupby("skill_threshold_met")
    .agg(
        job_count=("job_id", "nunique"),
        average_min_skill=("min_skill_level", "mean"),
        average_max_skill=("max_skill_level", "mean")
    )
    .reset_index()
)

skill_threshold_analysis["average_min_skill"] = (
    skill_threshold_analysis["average_min_skill"].round(2)
)

skill_threshold_analysis["average_max_skill"] = (
    skill_threshold_analysis["average_max_skill"].round(2)
)

print("\nSkill Threshold Analysis:")
print(skill_threshold_analysis)


# ============================================================
# STEP 19: Skill Threshold Percentage
# ============================================================

threshold_percentage = (
    df["skill_threshold_met"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .reset_index()
)

threshold_percentage.columns = [
    "skill_threshold_met",
    "percentage"
]

print("\nSkill Threshold Percentage:")
print(threshold_percentage)


# ============================================================
# STEP 20: Experience Requirement Analysis
# ============================================================

experience_analysis = (
    df.groupby("experience_required")
    .agg(
        job_count=("job_id", "nunique"),
        average_salary_lpa=("salary_lpa", "mean")
    )
    .reset_index()
    .sort_values("experience_required")
)

experience_analysis["average_salary_lpa"] = (
    experience_analysis["average_salary_lpa"].round(2)
)

print("\nExperience Requirement Analysis:")
print(experience_analysis)


# ============================================================
# STEP 21: Salary Analysis
# ============================================================

salary_analysis = pd.DataFrame({
    "metric": [
        "Minimum Salary",
        "Maximum Salary",
        "Average Salary",
        "Median Salary"
    ],
    "salary_lpa": [
        df["salary_lpa"].min(),
        df["salary_lpa"].max(),
        df["salary_lpa"].mean(),
        df["salary_lpa"].median()
    ]
})

salary_analysis["salary_lpa"] = (
    salary_analysis["salary_lpa"].round(2)
)

print("\nSalary Analysis:")
print(salary_analysis)


# ============================================================
# STEP 22: Minimum Skill Level Analysis
# ============================================================

minimum_skill_analysis = pd.DataFrame({
    "metric": [
        "Minimum Skill Level",
        "Maximum Skill Level",
        "Average Minimum Skill",
        "Median Minimum Skill"
    ],
    "skill_level": [
        df["min_skill_level"].min(),
        df["min_skill_level"].max(),
        df["min_skill_level"].mean(),
        df["min_skill_level"].median()
    ]
})

minimum_skill_analysis["skill_level"] = (
    minimum_skill_analysis["skill_level"].round(2)
)

print("\nMinimum Skill Level Analysis:")
print(minimum_skill_analysis)


# ============================================================
# STEP 23: Job Supply by Status and Location
# ============================================================

status_location_analysis = (
    df.groupby(["location", "job_status"])
    .agg(
        job_count=("job_id", "nunique")
    )
    .reset_index()
    .sort_values("job_count", ascending=False)
)

print("\nJob Supply by Status and Location:")
print(status_location_analysis)


# ============================================================
# STEP 24: Job Supply by Company and Status
# ============================================================

company_status_analysis = (
    df.groupby(["company_name", "job_status"])
    .agg(
        job_count=("job_id", "nunique")
    )
    .reset_index()
    .sort_values("job_count", ascending=False)
)

print("\nJob Supply by Company and Status:")
print(company_status_analysis)


# ============================================================
# STEP 25: Top Companies by Job Supply
# ============================================================

top_companies = (
    df.groupby("company_name")
    .agg(
        job_count=("job_id", "nunique")
    )
    .reset_index()
    .sort_values("job_count", ascending=False)
    .head(10)
)

print("\nTop Companies by Job Supply:")
print(top_companies)


# ============================================================
# STEP 26: Top Locations by Job Supply
# ============================================================

top_locations = (
    df.groupby("location")
    .agg(
        job_count=("job_id", "nunique")
    )
    .reset_index()
    .sort_values("job_count", ascending=False)
    .head(10)
)

print("\nTop Locations by Job Supply:")
print(top_locations)


# ============================================================
# STEP 27: Overall KPI Summary
# ============================================================

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Jobs",
        "Total Companies",
        "Active Jobs",
        "Closed Jobs",
        "Draft Jobs",
        "Average Salary LPA",
        "Average Experience Required",
        "Average Minimum Skill Level",
        "Successful Events",
        "Event Success Rate"
    ],
    "Value": [
        df["job_id"].nunique(),
        df["company_name"].nunique(),
        (df["job_status"] == "Active").sum(),
        (df["job_status"] == "Closed").sum(),
        (df["job_status"] == "Draft").sum(),
        round(df["salary_lpa"].mean(), 2),
        round(df["experience_required"].mean(), 2),
        round(df["min_skill_level"].mean(), 2),
        successful_events,
        round(event_success_rate, 2)
    ]
})

print("\nKPI Summary:")
print(kpi_summary)


# ============================================================
# STEP 28: Save All Analysis Results in One Excel File
# ============================================================

output_file = "Python Analysis Result.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned Data",
        index=False
    )

    kpi_summary.to_excel(
        writer,
        sheet_name="KPI Summary",
        index=False
    )

    job_status_analysis.to_excel(
        writer,
        sheet_name="Job Status",
        index=False
    )

    location_analysis.to_excel(
        writer,
        sheet_name="Location Analysis",
        index=False
    )

    company_analysis.to_excel(
        writer,
        sheet_name="Company Analysis",
        index=False
    )

    job_title_analysis.to_excel(
        writer,
        sheet_name="Job Title Analysis",
        index=False
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly Analysis",
        index=False
    )

    event_status_analysis.to_excel(
        writer,
        sheet_name="Event Status",
        index=False
    )

    event_success_summary.to_excel(
        writer,
        sheet_name="Event Success",
        index=False
    )

    source_analysis.to_excel(
        writer,
        sheet_name="Source Analysis",
        index=False
    )

    skill_threshold_analysis.to_excel(
        writer,
        sheet_name="Skill Threshold",
        index=False
    )

    threshold_percentage.to_excel(
        writer,
        sheet_name="Threshold Percentage",
        index=False
    )

    experience_analysis.to_excel(
        writer,
        sheet_name="Experience Analysis",
        index=False
    )

    salary_analysis.to_excel(
        writer,
        sheet_name="Salary Analysis",
        index=False
    )

    minimum_skill_analysis.to_excel(
        writer,
        sheet_name="Minimum Skill",
        index=False
    )

    status_location_analysis.to_excel(
        writer,
        sheet_name="Status Location",
        index=False
    )

    company_status_analysis.to_excel(
        writer,
        sheet_name="Company Status",
        index=False
    )

    top_companies.to_excel(
        writer,
        sheet_name="Top Companies",
        index=False
    )

    top_locations.to_excel(
        writer,
        sheet_name="Top Locations",
        index=False
    )

print("\nAnalysis Completed Successfully!")
print("All results saved in:", output_file)
```
